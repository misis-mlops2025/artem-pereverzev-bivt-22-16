```python
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
pdfmetrics.registerFont(TTFont('Arial', '/System/Library/Fonts/Supplemental/Arial Unicode.ttf'))
```

```python
from PyPDF2 import PdfReader, PdfWriter
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
from reportlab.lib.units import cm
from io import BytesIO

def image_with_caption_to_pdf(image_path, caption):
    buffer = BytesIO()
    c = canvas.Canvas(buffer, pagesize=A4)
    width, height = A4
    
    # Рисуем картинку (уменьшаем по высоте, чтобы осталось место для текста)
    img_height = height - 3 * cm
    c.drawImage(image_path, 0, 3 * cm, width, img_height, preserveAspectRatio=True, anchor='s')
    
    # Добавляем подпись под изображением
    c.setFont("Arial", 14)
    c.drawCentredString(width / 2, 1.5 * cm, caption)
    
    c.showPage()
    c.save()
    buffer.seek(0)
    return buffer
# пути
original_pdf = "/Users/artem.pereverzev/Downloads/Переверзев_Артем_Прак2-4.pdf"
output_pdf = "/Users/artem.pereverzev/Downloads/Переверзев_Артем_Прак2-4_with_add.pdf"
images = [
    ("/Users/artem.pereverzev/Documents/hw2.jpg", "Схема 1. Верхеуровневая структура предприятия."),
    ("/Users/artem.pereverzev/Documents/hw2-2.jpg", "Схема 2. Структура финансовой вертикали")
]

# читаем исходный PDF
reader = PdfReader(original_pdf)
writer = PdfWriter()

# добавляем все страницы оригинала
for page in reader.pages:
    writer.add_page(page)

# добавляем картинки как страницы
for img, cap in images:
    img_pdf = PdfReader(image_with_caption_to_pdf(img, cap))
    writer.add_page(img_pdf.pages[0])

# сохраняем
with open(output_pdf, "wb") as f:
    writer.write(f)
```
